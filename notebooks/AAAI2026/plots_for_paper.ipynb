### Code for plots to put in AAAI paper

```python
import matplotlib.pyplot as plt
import numpy as np


datasets = ['ETTh1', 'ETTh2', 'ETTm1', 'ETTm2', 'Exchange Rate', 'Traffic']
models = ['ZS', 'FFT', "Random Selection (ZS)", "Random Selection (FFT)", 'Our approach (using ZS)', 'Our approach (using FFT)']

mse_values = np.array([
    [0.0524, 0.0512, 0.0418, 0.0408, 0.0439, 0.0443],   
    [0.1306, 0.1304, 0.1038, 0.1039, 0.1002, 0.0992],   
    [0.0275, 0.0264, 0.0220, 0.0221, 0.0221, 0.0215],   
    [0.0765, 0.0646, 0.0615, 0.0518, 0.0644, 0.0555],  
    [0.0790, 0.0883, 0.0626, 0.0616, 0.0492, 0.0484],   
    [0.1766, 0.1173, 0.1412, 0.0936, 0.1407, 0.0933],    
])

# Create bar positions
x = np.arange(len(datasets))  
bar_width = 0.15

# Colors for models
colors = ['#1f77b4', '#ff7f0e', "#0AF3FF", '#9467bd', "#b4771f", '#2ca02c']

fig, ax = plt.subplots(figsize=(9, 5))

# Plot bars for each model
for i, model in enumerate(models):
    ax.bar(x + i * bar_width, mse_values[:, i], width=bar_width, label=model, color=colors[i])

# Axis labels and ticks
ax.set_xlabel('Dataset Name', fontsize=12)
ax.set_ylabel('MSE', fontsize=12)
# ax.set_title('Performance Comparison of Selective Forecasing using shapelets against baselines ', fontsize=13, pad=15)
ax.set_xticks(x + bar_width * (len(models) - 1) / 2)
ax.set_xticklabels(datasets, rotation=0, ha='center')

# Legend and grid
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.7)

# Optional: adjust layout
plt.tight_layout()
plt.show()
```

```python

datasets = ['ETTh1', 'ETTh2', 'ETTm1', 'ETTm2', 'Exchange Rate', 'Traffic']
drop_percent = [10, 20, 30, 40, 50]



mse_values = np.array([
    [0.0482, 0.0442, 0.0400, 0.0344, 0.0312],   
    [0.1126, 0.0992, 0.0872, 0.0758, 0.0654],   
    [0.0253, 0.0215, 0.0195, 0.0165, 0.0133 ],   
    [0.0607, 0.0555, 0.0484, 0.0413, 0.0341],  
    [0.0622, 0.0484, 0.0375, 0.0293, 0.0266],   
    [0.1031, 0.0933, 0.0787, 0.0669, 0.0567],    
])




# Plot all datasets
plt.figure(figsize=(8, 5))
for i, dataset in enumerate(datasets):
    plt.plot(drop_percent, mse_values[i], marker='o', linestyle='--', label=datasets[i])
# for label, errors in datasets.items():
#     plt.plot(drop_percent, errors, marker='o', linestyle='-', label=label)

# Labeling and legend
plt.xlabel("Drop Percentage (%)")
plt.ylabel("MSE")
# plt.title("Error vs Drop Percent for Multiple Datasets")
plt.grid(True)
plt.legend(title="Dataset")

# Optional: Save plot
plt.savefig("error_vs_drop_multiple.png", dpi=150, bbox_inches="tight")

plt.show()
```

```python
#  percentage improvement

improvements = (mse_values[:, 3] - mse_values[:, 5]) / mse_values[:, 3] * 100
for i, dataset in enumerate(datasets):
    print(f"Percentage improvement for {dataset}: {improvements[i]:.2f}%")
mean_improvement = np.mean(improvements)
print(f"Mean Percentage improvement across datasets: {mean_improvement:.2f}%")
```
